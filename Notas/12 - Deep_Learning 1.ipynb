# Un poco de historia

Antes de comenzar con la explicación de qué es el aprendizaje profundo, demos un vistazo a la historia general de la inteligencia artificial (IA). Podemos dividirla en periodos de mucho entusiasmo general sobre el tema (*boom*) seguidos de periodos en los que el interés general de las personas decrece significativamente (*winters*). 

## Primer Boom

El primer auge de la IA comenzó alrededor de 1956, con la Conferencia de Dartmouth, donde John McCarthy, Marvin Minsky, Allen Newell y Herbert Simon propusieron el término Inteligencia Artificial. En esa época, los primeros computadores digitales permitieron implementar programas capaces de resolver problemas de manera automática mediante búsquedas informadas y no informadas.

Entre los hitos más importantes de este periodo se encuentran el Logic Theorist (1955) y el General Problem Solver (1959) de Newell y Simon, que empleaban estrategias de análisis medios-fines (means–ends analysis) para acercarse al objetivo desde el estado actual. En IBM, Arthur Samuel desarrolló uno de los primeros programas que aprendía a jugar damas (checkers) mediante auto-juego, anticipando conceptos de lo que hoy conocemos como aprendizaje por refuerzo.

Estos éxitos inicales llevaron a la agencia DARPA de Estados Unidos (*Defense Advanced Research Projects Agency*) a dedicar grandes sumas de dinero a los proyectos de IA de diferentes universidades. El optimismo era enorme: se pensaba que las máquinas alcanzarían pronto la inteligencia humana. De hecho, Marvin Minsky llegó a afirmar en una entrevista de 1970 que “dentro de tres a ocho años tendremos una máquina con la inteligencia de un ser humano promedio”. Las personas incluso empezaron a temerle a la IA, entre 1953 y 1964, cientos de artículos se publicaron en Estados Unidos sobre la influencia de la tecnología en la eliminación de trabajos.

Sin embargo, las limitaciones pronto se hicieron evidentes. El Perceptrón de Rosenblatt (1958), uno de los primeros modelos de red neuronal implementado incluso en hardware, resultó incapaz de resolver problemas no linealmente separables. El libro Perceptrons de Minsky y Papert (1969) mostró rigurosamente esas limitaciones.

Además, en 1973, James Lighthill publicó un artículo titulado *"Artificial Intelligence: A General Survey"*, que se conoce popularmente como "el reporte Lighthill". Este reporte hacía un pronóstico muy pesimista de muchos aspectos centrales del área, diciendo que "en ninguna parte del campo se han hecho descubrimientos hasta el momento, que hayan producido el impacto que prometían". Este informe es un reflejo de la visión de la época y, sumado a DARPA suspendiendo la financiación de proyectos, se llegó al primer invierno IA entre 1974 y 1980.


## Segundo Boom

El segundo auge de la IA comenzó a finales de los años 70 con la aparición de los sistemas expertos.
A diferencia de los agentes lógicos generales del primer periodo, estos sistemas estaban diseñados para resolver problemas dentro de un dominio específico, combinando una base de conocimiento con un motor de inferencia.

Ejemplos destacados fueron MYCIN, desarrollado en la Universidad de Stanford para recomendar antibióticos mediante encadenamiento hacia atrás (backward chaining), y DENDRAL, usado en química orgánica para identificar estructuras moleculares. Posteriormente, CADUCEUS (antes INTERNIST-I) se convirtió en uno de los sistemas de diagnóstico más complejos de la época, capaz de manejar cientos de enfermedades. También surgieron lenguajes especializados como Prolog (1972) y las Lisp Machines del MIT, diseñadas para ejecutar programas de IA con mayor eficiencia.

Como en el primer boom, las expectativas fueron desmesuradas: se pensaba que bastaba con “codificar el conocimiento” para obtener inteligencia. Pero mantener estas bases de conocimiento era costoso, difícil de actualizar y dependía de expertos humanos. Hacia 1987, la caída de las empresas dedicadas a Lisp Machines y la reducción de los fondos de investigación marcaron el segundo invierno de la IA.

## Tercer Boom

A finales de los 80 surgieron avances que sentaron las bases del auge actual. El algoritmo de retropropagación del error (backpropagation), descrito por Rumelhart, Hinton y Williams (1986), permitió entrenar de forma práctica las redes neuronales multicapa (MLP). En paralelo, Vladimir Vapnik y Corinna Cortes (1992) desarrollaron las máquinas de vectores de soporte (SVM), que aprovecharon el truco del kernel para extender los modelos lineales a espacios de alta dimensión. Durante los años 90, estas técnicas dominaron el aprendizaje automático, aunque las redes neuronales seguían limitadas por el poder de cómputo y los datos disponibles.

El aprendizaje profundo surge cuando una ANN tiene más de una capa oculta. En términos simples, una red neuronal profunda es aquella con varias capas ocultas intermedias entre la entrada y la salida. Cada capa aprende representaciones más abstractas del dato, de modo que las primeras detectan patrones simples (bordes, frecuencias, palabras), y las últimas capturan estructuras de mayor nivel (formas, objetos, significados). Así, el aprendizaje profundo permite que el modelo aprenda directamente las características relevantes, reduciendo la necesidad de diseñarlas manualmente. 

En el año 1999 Nvidia popularizó las GPU (término que ya se usaba desde 1994 para las playstation), pero no fue hasta 2011 cuando empezaron a usarse para entrenar redes neuronales. El punto de inflexión fue 2012, cuando la red AlexNet (Krizhevsky, Sutskever y Hinton) ganó la competencia ImageNet reduciendo drásticamente el error de clasificación. Desde entonces, el aprendizaje profundo se convirtió en el paradigma dominante en visión por computador, procesamiento de lenguaje natural, voz, recomendadores y casi cualquier área de la IA actual.


El aprendizaje profundo (deep learning) resurgió cuando se combinaron tres factores:

- Mayor capacidad de cómputo gracias a las GPU programables (CUDA, 2006).

- Grandes volúmenes de datos disponibles en internet.

- Mejoras en regularización y optimización (Dropout, ReLU, Batch Normalization).

# Redes neuronales convolucionales

La convolución es una operación matemática que transforma dos funciones $f$ y $g$ en una tercera función que representa la magnitud en la que se superponen $f$ y $g$. Se define como:

\begin{equation}
f(t) * g(t) = \sum_a f^T(t)g(t-a)
\end{equation}

Veamos un ejemplo de convolución: 

```python
import numpy as np
import matplotlib.pyplot as plt
f=np.hstack([np.repeat(0,5),np.repeat(5,5),np.repeat(0,5)])
g=np.hstack([np.repeat(0,5),np.arange(5,0,-1),np.repeat(0,5),np.arange(0,5,1),np.repeat(0,5),np.repeat(5,5),np.repeat(0,5),np.arange(0,5,1),np.repeat(0,5)])
plt.subplot(211)
plt.plot(f)
plt.subplot(212)
plt.plot(g)
plt.show()
```

```python
h=np.zeros(len(g)-len(f))
for a in range(len(h)):
    h[a]=f.dot(g[a:(a+len(f))])
plt.subplot(211)
plt.plot(h)
plt.subplot(212)
plt.plot(g,c='red')
```

A la señal más corta se le denomina *kernel* (no confundirlo con los kernels en SVM) y se realizará la convolución de éste con la señal de entrada.

# Dropout: aprender sin memorizar

Uno de los mayores riesgos en redes neuronales profundas es que aprendan demasiado bien los datos de entrenamiento — es decir, que memoricen los ejemplos en lugar de aprender patrones generales. A eso se le llama sobreajuste (overfitting).

Una técnica muy efectiva para evitarlo es *Dropout*, propuesta por Geoffrey Hinton en 2014.

Durante el entrenamiento, Dropout “apaga” aleatoriamente una fracción de las neuronas de una capa (por ejemplo, el 20% o el 50%). Esto significa que, en cada paso, la red se entrena con una versión ligeramente diferente de sí misma: cada iteración ve una subred distinta. Ninguna neurona puede “depender demasiado” de las otras y así la red aprende representaciones más robustas y distribuidas.

Al final, durante la inferencia, todas las neuronas vuelven a estar activas, pero sus salidas se escalan para compensar los apagones del entrenamiento.

# Funciones de pérdida en las redes neuronales

Ahora vamos a ver la cómo funcionan los diferentes algoritmos de entrenamiento para las redes neuronales profundas. Primero recordemos lo que habíamos visto para las redes *feedforward*, cuyo entrenamiento se basa en el gradiente descendente estocástico (SGD):

\begin{equation}
\boldsymbol{w}^{(\tau+1)} = \boldsymbol{w}^{(\tau)} - \alpha\nabla \varepsilon_p
\end{equation}

donde $\varepsilon_p$ es el error cometido por el clasificador para la muestra $p$. Usando mínimos cuadrados, tenemos:

\begin{equation}
\varepsilon_p = \frac{1}{2}\sum_k (s_{k,p} - y_{k,p})^2
\end{equation}

donde $s_{k,p}$ es la k-ésima salida de la red para la muestra $p$ y $y_{k,p}$ es el objetivo correspondiente (etiqueta).

Ahora, esta definición de error no es única, sino que existen muchas formas de medir qué tan lejos están las predicciones hechas por la red, de los objetivos propuestos. Estas formas de medición se denominan "funciones de pérdida" y su definición puede facilitar el algoritmo de entrenamiento. Actualmente, la función de pérdida más comúnmente usada para los problemas de clasificación, es la entropía cruzada. Veamos en qué consiste.

## Entropía cruzada binaria

La definición de entropía cruzada binaria proviene del uso de la estimación de máxima verosimilitud (teorema de Bayes) para una variable aleatoria con distribución de probabilidad de Bernoulli. La distribución de Bernoulli (también llamada distribución dicotómica), es una distribución de probabilidad discreta, que toma valor $p$ para la probabilidad de éxito y valor $q=1-p$ para la probabilidad de fracaso. 

\begin{equation}
p(a) = p^a(1-p)^{1-a}
\end{equation}

En este sentido, para un problema de clasificación binaria donde las muestras las denominamos $\boldsymbol{x}_i$, $\forall i=1\ldots N$, podemos interpretar las ettiquetas de clase $y_i \in {0,1}$ como una variable aletoria que nos indica la probabilidad de que se "acierte" o se "fracase" en la clasificación. Supongamos que tenemos algún modelo de clasificación que nos entrega una predicción binaria $c_i$, interpretada como la probabilidad de pertenecer a una clase o la otra y que depende de un vector de parámetros $\boldsymbol{w}$:

\begin{equation}
c_i = f(\boldsymbol{x}_i , \boldsymbol{w})
\end{equation}

La distribución de probabilidad de las muestras (suponiendo independencia) será entonces la multiplicación de todas las probabilidades, condicionada por $\boldsymbol{w}$:

\begin{equation}
p(\boldsymbol{X},\boldsymbol{y}) = \prod_{i=1}^N c_i^{y_i} (1 -  c_i)^{(1-y_i)}
\end{equation}

\begin{equation}
p(\boldsymbol{X},\boldsymbol{y} | \boldsymbol{w}) = \prod_{i=1}^N f(\boldsymbol{x}_i , \boldsymbol{w})^{y_i} (1 -  f(\boldsymbol{x}_i , \boldsymbol{w}))^{(1-y_i)}
\end{equation}

La idea es encontrar el vector de parámetros $\boldsymbol{w}$ que maximice esta probabilidad (verosimilitud). Para ello, podemos aplicar logaritmo que nos convierte las multipicaciones en sumas y no cambia la ubicación del máximo, obteniendo la log-verosimilitud $\mathcal{L}(\boldsymbol{X},\boldsymbol{y} | \boldsymbol{w})$:

\begin{equation}
\mathcal{L}(\boldsymbol{X},\boldsymbol{y} | \boldsymbol{w}) = \sum_{i=1}^N y_i \operatorname{log}(f(\boldsymbol{x}_i , \boldsymbol{w}))(1-y_i) \operatorname{log}(1 -  f(\boldsymbol{x}_i , \boldsymbol{w}))
\end{equation}

Como la mayoría de paquetes de optimización están hechos para **minimizar**, invertimos la función multiplicándola por $-1$ y esta función se conoce como entropía cruzada binaria:

\begin{equation}
H = -\sum_{i=1}^N y_i \operatorname{log}(f(\boldsymbol{x}_i , \boldsymbol{w}))(1-y_i) \operatorname{log}(1 -  f(\boldsymbol{x}_i , \boldsymbol{w}))
\end{equation}

## Entropía cruzada

En el caso multiclase, ahora las etiquetas se asignan a una cantidad arbitraria de $m$ valores $y_i \in {0,1,2,\ldots,m}$. E este caso, en lugar de las distribución de Bernoulli, tenemos una distribución binomial:

\begin{equation}
p(\boldsymbol{X},\boldsymbol{y} | \boldsymbol{w}) = \frac{N!}{N_1! N_2! \ldots N_m!}\prod_{i=1}^N f(\boldsymbol{x}_i , \boldsymbol{w})^{y_i}
\end{equation}

donde $N_i$ es la cantidad de datos de la i-ésima clase. Siguiendo el mismo procdimiento, llegamos a:

\begin{equation}
\mathcal{L}(\boldsymbol{X},\boldsymbol{y} | \boldsymbol{w}) = \operatorname{log}(N!)-\operatorname{log}(N_1!)-\operatorname{log}(N_2!)-\ldots-\operatorname{log}(N_m!)+\sum_{i=1}^N y_i \operatorname{log}(f(\boldsymbol{x}_i , \boldsymbol{w}))
\end{equation}

Sin embargo, como los términos con factoriales de $N_i$ no dependen de $\boldsymbol{w}$, van aser constantes en el proceso y los podemos eliminar. LLegamos entonces a la fórmula de la entropía cruzada (compararla con la entropía de Shannon):

\begin{equation}
H = -\sum_{i=1}^N y_i \operatorname{log}(f(\boldsymbol{x}_i , \boldsymbol{w}))
\end{equation}

# Funciones de activación

Ahora analicemos las funciones de activación más comúnmente usadas en las redes neuronales profundas. Recordemos que estas funciones se usan a la salida de cada neurona para realizar mapeos no lineales:

![](perceptron.png)

Las funciones de activación deben ser continuas para que sean diferenciables y poder aplicar el gradiente descendente. Ya habímos hablado de las funciones sigmoidal y tangente hiperbólica:

\begin{equation}
\sigma(x) = \frac{1}{1+e^{-x}}
\end{equation}

```python
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(-100,100,200)
def sigm(x):
    y = 1/(1+np.exp(-x))
    return(y)
sigmoidal=sigm(x)
tangente=np.tanh(x)

#Gráfica
plt.rcParams["figure.figsize"]=20,5 #Modifica el tamaño de la gráfica
plt.subplot(1,2,1)
plt.plot(x,tangente)
plt.title('Tangente hiperbólica')
plt.subplot(1,2,2)
plt.plot(x,sigmoidal)
plt.title('Sigmoide')
plt.show()
```

Como vemos, estas funciones de activación mapean toda el eje real a los valores discretos 0 o 1. Por lo tanto, son las que más se usan para la capa de salida en problemas de clasificación biclase usando (una sola neurona de salida) o para la salida de problemas multiclase usando *hot encoding* (una neurona para cada clase). También se usan en las capas ocultas de redes totalmente conectadas, donde simulan la sinapsis neuronal.

Otra función comúnmente usada es la *softmax*. Esta función es una normalización de toda la capa de manera que las salidas de las capas anteriores sumen uno. Típicamente, las neuronas de las capas anteriores entregan un valor no normalizado de qué tan viable es que las entradas pertenezcan a una clase en particular. La capa con funciones softmax normaliza estos valores de manera que la salida represente una probabilidad para cada clase: 

\begin{equation}
\sigma_i(\boldsymbol{a}) = \frac{e^{a_i}}{\sum_j e^{a_j}}
\end{equation}

```python
#Supongamos que la capa anterior entrega valores no normalizados
entrada=np.array([2,1,0.1])
#La salida softmax será entonces:
softmax = np.exp(entrada) / sum(np.exp(entrada))
print(entrada)
print(softmax)
```

Una de las funciones de activación que más fuerza ha tomado con el aprendizaje profundo es la Unidad Lineal Rectificada (ReLU). La razón recae en un problema conocido como "desvanecimiento del gradiente". 

Recordemos que el algoritmo de entrenamiento de las redes neuronales está basado en el uso del gradiente descendente estocástico. El algoritmo primero calcula el error de predicción cometido por el modelo (función de pérdida) y luego lo usa para estimar un gradiente que se emplea en la actualización de todos los pesos de la red de manera que el error sea menor en la próxima iteración. La clave para entender el desvanecimiento del gradiente radica en que este gradiente **se propaga hcia atrás a través de toda la red** desde la capa de salida hasta la capa de entrada. Cuando usamos las funciones de activación acotadas como las que hemos visto hasta ahora, **el error se acota múltiples veces y puede llegar a ser tan pequeño al llegar a las primeras capas que su efecto sea despreciable**.

Este efecto indeseable se hace más patente cuantas más capas tenga la red, por lo cual fue precisamente con el surgimiento del aprendizaje profundo cuando se hizo claro que era necesario usar funciones de activación no acotadas. La función ReLU es simplemente una "rectificación" de la función lineal, es decir, una función que es cero para valores negativos y lineal para valores positivos. Esta función entonces no es acotada y conserva las propiedades deseables de la función lineal, al mismo tiempo que tiene una no linealidad que permite que la red aprenda sobre datos más complejos.

\begin{equation}
\sigma(x) = \operatorname{max}(0,x)
\end{equation}

```python
x=np.linspace(-100,100,200)
def ReLU(x):
    y = np.maximum(np.repeat(0,len(x)),x)
    return(y)
RL=ReLU(x)
#Gráfica
plt.plot(x,RL)
plt.title('ReLU')
plt.show()
```

Resumiendo:

**Problemas de regresión** (predicción de valores reales):
* Configuración de la capa de salida: Una sola neurona con función de activación lineal
* Función de pérdida: Mínimos cuadrados

**Problemas de clasificación binaria:**
* Configuración de la capa de salida: Una sola neurona con función de activación sigmoidal
* Función de pérdida: Entropía cruzada binaria

**Problemas de clasificación multiclase:**
* Configuración de la capa de salida: una neurona para cada clase con función de activación softmax
* Función de pérdida: Entropía cruzada

Para todos en las capas ocultas: Si el modelo es profundo, usar ReLU. Si el modelo es superficial, usar sigmoide o tanh.

```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
#Usaremos un conjunto de datos llamado  CIFAR10 que contiene 60,000 imágenes en color en 10 clases. 
#50.000 se usan para entrenar y 10.000 para probar
(train_images, train_labels), (test_images, test_labels)=tf.keras.datasets.cifar10.load_data()

#Normalización
train_images = train_images.astype("float32") / 255.0
test_images = test_images.astype("float32") / 255.0
```

```python
#Veamos algunos ejemplos
class_names = ['airplane', 'automobile', 'bird', 'cat', 'deer',
               'dog', 'frog', 'horse', 'ship', 'truck']

#Las figuras originales son de 32 X 32 pero las voy a mostrar en 10 X 10
plt.figure(figsize=(10,10))
for i in range(50):
    plt.subplot(5,10,i+1)
    plt.xticks([])
    plt.yticks([])
    plt.grid(False)
    plt.imshow(train_images[i], cmap=plt.cm.binary)
    #plt.xlabel(class_names[train_labels[i][0]])
plt.show()
```

Ahora definiremos la arquitectura de la red:
1. Vamos a aumentar los datos, introduciendo *flips* horizontales aleatorios y rotaciones del 10%.
2. Nuestros objetos de entrada son tensores de tamaño 32 X 32 X 3 (R,G,B), de manera que la capa de entrada debe tener estas dimensiones. 
3. Definiremos el número de neuronas de la capa (filtros) y el tamaño del kernel (en este caso usaremos un kernel de 3 X 3, lo que significa que cada neurona se conecta con 9 neuronas de la siguiente capa).
2. Usaremos la función de activación ReLu en las 5 capas convolucionales, dos de 32 neuronas, dos de 64 neuronas y una de 128 neuronas.
3. Cada dos capas convolucionales pondremos una capa de *max pooling* con un elemento estructurante de 2 X 2.

```python
modelo = models.Sequential()
#Capa de entrada
modelo.add(layers.Input(shape=(32,32,3)))
#Data augmentation
modelo.add(layers.RandomFlip("horizontal"))
modelo.add(layers.RandomRotation(0.1))
#Capas convolucionales y pooling
modelo.add(layers.Conv2D(32, (3, 3), activation='relu'))
modelo.add(layers.Conv2D(32, (3, 3), activation='relu'))
modelo.add(layers.MaxPooling2D((2, 2)))
modelo.add(layers.Conv2D(64, (3, 3), activation='relu'))
modelo.add(layers.Conv2D(64, (3, 3), activation='relu'))
modelo.add(layers.MaxPooling2D((2, 2)))
modelo.add(layers.Conv2D(128, (3, 3), activation='relu'))
layers.MaxPooling2D((2, 2))

#Veamos el resumen de la red hasta el momento:
modelo.summary()
```

Finalmente, agregaremos capas totalmente conectadas que son las que en realidad hacen la clasificación. Estas capas no reciben tensores tridimensionales sino que es necesario aplanar los datos antes de que ingresen:

```python
modelo.add(layers.Flatten())
modelo.add(layers.Dense(256, activation='relu'))
modelo.add(layers.Dropout(0.5))
modelo.add(layers.Dense(10,activation='softmax'))
modelo.summary()
```

Ahora entrenamos el modelo usando el algoritmo Gradiente Descendente Estocástico (SGD), con una función de pérdida basada en la entropía cruzada.Probemos varias tasas de aprendizaje. 

```python
optimizador = 'adam'#tf.keras.optimizers.SGD(learning_rate=0.0001)
modelo.compile(optimizer=optimizador,
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=False),
              metrics=['accuracy'])

history=modelo.fit(train_images, train_labels, epochs=20, batch_size=64,
                    validation_data=(test_images, test_labels))
```

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('Model Accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper left')
plt.show()
```

```python
#Ahora vamos a validar el entrenamiento sobre los datos de prueba
test_loss, test_acc = modelo.evaluate(test_images,  test_labels, verbose=2)
print(test_acc)
```
